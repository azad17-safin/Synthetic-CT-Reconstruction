```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.data import shepp_logan_phantom
from skimage.transform import radon, iradon

image = shepp_logan_phantom()
plt.figure(figsize=(6, 6))
plt.title("Original Phantom Image (Simulated Brain)")
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.show()

theta = np.linspace(0., 180., max(image.shape), endpoint=False)
sinogram = radon(image, theta=theta)

plt.figure(figsize=(8, 6))
plt.title("Sinogram (Raw CT Detector Data)")
plt.xlabel("Projection Angle (degrees)")
plt.ylabel("Sensor Position (pixels)")
plt.imshow(sinogram, cmap='gray', aspect='auto')
plt.colorbar(label="X-ray Absorption")
plt.show()

reconstructed_image = iradon(sinogram, theta=theta, filter_name='ramp')

plt.figure(figsize=(6, 6))
plt.title("Reconstructed CT Image (Inverse Radon Transform)")
plt.imshow(reconstructed_image, cmap='gray')
plt.axis('off')
plt.show()
```
